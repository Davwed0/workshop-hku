## Setup: Import Libraries

First, we import the libraries we'll use. Think of this like opening Excel.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display all columns
pd.set_option('display.max_columns', None)

print("✅ Libraries loaded successfully!")
```

## 1. Loading Data

### Excel Way:
- File → Open → Browse for file
- Double-click to open

### Pandas Way:

```python
# Load the data (like opening an Excel file)
df = pd.read_csv('../data/superstore_sales.csv')

print("Data loaded successfully!")
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
```

## 2. Viewing Data

### Excel Way:
- Scroll through the spreadsheet
- Click on cells to view values

### Pandas Way:

```python
# View first few rows (like scrolling to the top)
df.head()
```

```python
# View last few rows
df.tail()
```

```python
# Get basic information about the data
df.info()
```

## 3. Basic Statistics

### Excel Way:
- Use functions like `=SUM()`, `=AVERAGE()`, `=COUNT()`
- Insert → Quick Analysis → Totals

### Pandas Way:

```python
# Get summary statistics for all numeric columns
df.describe()
```

### Side-by-Side Comparison: SUM Function

```python
# Excel: =SUM(D2:D61)
# Pandas:
total_sales = df['Sales'].sum()
print(f"Total Sales: ${total_sales:,.2f}")
```

### Side-by-Side Comparison: AVERAGE Function

```python
# Excel: =AVERAGE(D2:D61)
# Pandas:
avg_sales = df['Sales'].mean()
print(f"Average Sales: ${avg_sales:,.2f}")
```

### Side-by-Side Comparison: COUNT Function

```python
# Excel: =COUNT(D2:D61)
# Pandas:
count_sales = df['Sales'].count()
print(f"Number of Sales Records: {count_sales}")
```

### Side-by-Side Comparison: MIN and MAX

```python
# Excel: =MIN(D2:D61) and =MAX(D2:D61)
# Pandas:
min_sales = df['Sales'].min()
max_sales = df['Sales'].max()
print(f"Minimum Sales: ${min_sales:,.2f}")
print(f"Maximum Sales: ${max_sales:,.2f}")
```

## 4. Filtering Data

### Excel Way:
- Data → Filter → Click dropdown → Select values

### Pandas Way:

```python
# Filter for Laptop sales only
laptop_sales = df[df['Product'] == 'Laptop']
print(f"Laptop sales records: {len(laptop_sales)}")
laptop_sales.head()
```

```python
# Filter for sales above $500
high_value_sales = df[df['Sales'] > 500]
print(f"High value sales (>$500): {len(high_value_sales)}")
high_value_sales.head()
```

```python
# Multiple conditions: Laptops in North region
north_laptops = df[(df['Product'] == 'Laptop') & (df['Region'] == 'North')]
print(f"Laptop sales in North region: {len(north_laptops)}")
north_laptops
```

## 5. Creating New Columns

### Excel Way:
- Add a new column header
- Write formula like `=D2-E2`
- Drag formula down

### Pandas Way:

```python
# Calculate Profit (Sales - Cost)
# Excel: =D2-E2 (in cell F2, then drag down)
# Pandas:
df['Profit'] = df['Sales'] - df['Cost']

# Calculate Profit Margin as percentage
# Excel: =(D2-E2)/D2*100
# Pandas:
df['Profit_Margin'] = (df['Profit'] / df['Sales']) * 100

df.head()
```

## 6. Sorting Data

### Excel Way:
- Select data → Data → Sort → Choose column and order

### Pandas Way:

```python
# Sort by Sales (descending)
df_sorted = df.sort_values('Sales', ascending=False)
print("Top 10 sales:")
df_sorted.head(10)
```

```python
# Sort by multiple columns
df_multi_sort = df.sort_values(['Region', 'Sales'], ascending=[True, False])
print("Sorted by Region, then Sales (descending):")
df_multi_sort.head(10)
```

## 7. Simple Charts

### Excel Way:
- Select data → Insert → Chart → Choose type

### Pandas Way:

```python
# Sales by Product - Bar Chart
product_sales = df.groupby('Product')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
product_sales.plot(kind='bar', color='steelblue')
plt.title('Total Sales by Product', fontsize=14, fontweight='bold')
plt.xlabel('Product', fontsize=12)
plt.ylabel('Sales ($)', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Sales by Region - Pie Chart
region_sales = df.groupby('Region')['Sales'].sum()

plt.figure(figsize=(8, 8))
plt.pie(region_sales, labels=region_sales.index, autopct='%1.1f%%', startangle=90)
plt.title('Sales Distribution by Region', fontsize=14, fontweight='bold')
plt.axis('equal')
plt.tight_layout()
plt.show()
```
